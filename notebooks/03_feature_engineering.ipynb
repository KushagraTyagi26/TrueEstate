```python
import pandas as pd

df = pd.read_csv('../data/raw/cities_magicbricks_rental_prices.csv')

df.head()
```

```python
print(df.dtypes)
```

```python
print("House Types:\n")
print(df['house_type'].value_counts())

print("\nFurnishing Types:\n")
print(df['furnishing'].value_counts())

print("\nCities:\n")
print(df['city'].value_counts())
```

```python
numeric_columns = ['area', 'beds', 'bathrooms', 'balconies', 'area_rate', 'rent']

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print(df[numeric_columns].dtypes)
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=['house_type', 'city', 'furnishing'],
    drop_first=True
)

print(df_encoded.head())
print()
print("Encoded dataset shape:", df_encoded.shape)
```

```python
# Area available per bedroom
df_encoded['area_per_bedroom'] = df_encoded['area'] / df_encoded['beds']

# Bathrooms per bedroom
df_encoded['bath_per_bedroom'] = df_encoded['bathrooms'] / df_encoded['beds']

# Balconies per bedroom
df_encoded['balcony_per_bedroom'] = df_encoded['balconies'] / df_encoded['beds']

print(df_encoded[['area', 'beds', 'area_per_bedroom',
                  'bath_per_bedroom', 'balcony_per_bedroom']].head())
```

```python
print(df_encoded.isnull().sum().sum())
print()
print(df_encoded.isnull().sum()[df_encoded.isnull().sum() > 0])
```

```python
df_encoded = df_encoded.dropna()

print("Final dataset shape:", df_encoded.shape)
```

```python
df_encoded.to_csv('../data/processed/rental_processed.csv', index=False)

print("Processed dataset saved successfully!")
```
